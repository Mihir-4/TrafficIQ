```python
from pathlib import Path
import os
import random
import yaml
import numpy as np
import torch

from ultralytics import YOLO

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

PROJECT_ROOT = Path("/kaggle/working/TrafficIQ")

DATASET_YAML = PROJECT_ROOT / "configs" / "dataset.yaml"

MODELS_DIR = PROJECT_ROOT / "models"
RESULTS_DIR = PROJECT_ROOT / "training"

MODELS_DIR.mkdir(exist_ok=True)
RESULTS_DIR.mkdir(exist_ok=True)

print(DATASET_YAML)
```

```python
with open(DATASET_YAML) as f:
    config = yaml.safe_load(f)

print(config)

assert os.path.exists(DATASET_YAML)

print("\nDataset configuration loaded successfully.")
```
